```python
x = 100
print(x)
print(x/0)
y = 'Hello'
print(y)

# Now we shall do exception handling using try except block
```

```python
try:
    x = 100
    print(x)
    print(x/0)
    y = 'Hello'
    print(y)
except:
    print('There is some error')

# Firstly try block will be executed by default, once any error is encountered it switched to except block
```

```python
try:
    x = 100
    print(x)
    print(x/0)
    y = 'Hello'
    print(y)
except Exception as e:
    print(e)
```

```python
try:
    x = 100
    print(x)
    print(x/0)
    y = 'Hello'
    print(y)
except ZeroDivisionError as z:
    print(z)
except TypeError as t:
    print(t)
```

```python
try:
    x = 100
    print(x)
    print(x/0)
    y = 'Hello'
    print(y)
except Exception as e:
    print(e)
finally:
    print('This run always')
```
