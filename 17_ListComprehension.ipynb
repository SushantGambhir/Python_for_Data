```python
l = [1,2,3,4,5]
```

```python
dbl = []

for i in l:
    dbl.append(2*i)
print(dbl)
```

```python
# The above code could be done alternatively via list comprehension

dbl2 = [2*i for i in l]
print(dbl2)
```

```python
coun = ['india','usa','uk']

countries = [j.upper() for j in coun]

print(countries)
```

```python
# Squares of first 20 integers

sqlist = [i*i for i in range(1,21)]
print(sqlist)
```
