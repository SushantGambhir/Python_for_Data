```python
# Problem statement: Remove duplicate items from a list
l = [4,1,2,3,4,1,1,2,2,3,4]
s = set(l)
l_ = list(s)
print(l_)
```

```python
# Set union
s1 = {1,2,3,4}
s2 = {5,6,7,8}
print(s1.union(s2))
```

```python
# Set difference
s3 = {1,2,4,5}
s4 = {2,1,7}

print(s3.difference(s4)) # s3 - s4
print(s4.difference(s3)) # s4 - s3
```
