```python
# Dictionaries have key value pairs
empdt = {'ID':1,'Name':'John','Salary':10000}
print(empdt)
type(empdt)
```

```python
# We dont use indexing, instead we use keys
# Dictionaries are mutable

print(empdt['ID'])
```

```python
empdt['City'] = 'New York City'
print(empdt)
```

```python
print('City' in empdt)
```

```python
# Looping in dictionaries

for i in empdt.keys():
    print(i)

# Keys will be printed
```

```python
for i in empdt.values():
    print(i)

# Values will be printed
```

```python
for i,j in empdt.items():
    print(i,j)

# Both key and values will be printed
```
