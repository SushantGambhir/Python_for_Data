```python
x = 'Hello'
print(x.find('e')) #Gives the index of 'e'
print(x.find('l')) #Gives the index of first 'l'
print(x.find('z')) #Returns -1 because not present in string
print(x.find('l',3)) # Returns first instance of l starting from index 3
```
