```python
# Map applies a given function to every item in an iterable (like a list or tuple) and returns an iterator of the results.
```

```python
lst = [1,2,3,4,5]
lst
```

```python
def doublenum(x):
    return 2*x
```

```python
mp = map(doublenum,lst)
type(mp)
```

```python
mp
```

```python
list(mp)
```

```python
# - Once you consume a map object (e.g., by converting it to a list or looping through it), it’s exhausted. You can’t reuse it unless you call map() again.
```

```python
countries = ['india','usa','uk']
```

```python
def capital(x):
    return x.upper()
```

```python
cmp = list(map(capital,countries))
```

```python
cmp
```
