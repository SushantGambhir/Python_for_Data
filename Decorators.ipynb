A decorator in Python is a special function that modifies the behavior of another function without changing its
code. It allows you to dynamically enhance functions or methods.


Enhances functions dynamically


Used for logging, authentication, caching, timing, etc.


Follows the principle of higher-order functions (functions inside functions)

```python
def greetings():
    print('Hello World')
```

```python
greetings()
```

```python
def convert_to_uppercase(func):
    def inner():
        result = func() # Calls the original function
        return result.upper() #Modify the result
    return inner # Return the modified function
```

```python
@convert_to_uppercase

def greet():
    return 'hello world'

greet()
```

```python
def square_number(func):
    def inner(a):
        i = func(a)
        return i*i
    return inner
```

```python
@square_number

def numbers(a):
    return a
```

```python
print(numbers(2))
```

```python
def square_all_numbers(func):
    def inner(a):
        x = func(a)
        return [i*i for i in x] 
    return inner
```

```python
@square_all_numbers
def test_lst(x):
    return x

test_lst([1,2,3,4,5])
```

```python
def cube_numbers(func):
    def inner(a):
        i = func(a)
        return i**3
    return inner
```

```python
@square_number
@cube_numbers
def numbers4(x):
    return x

print(numbers4(2))

# The lower ones apply first (here cube happens and then square)
```

```python
@cube_numbers
@square_number
def numbers4(x):
    return x

print(numbers4(2))
```
