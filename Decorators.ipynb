A decorator in Python is a special function that modifies the behavior of another function without changing its
code. It allows you to dynamically enhance functions or methods.


Enhances functions dynamically


Used for logging, authentication, caching, timing, etc.


Follows the principle of higher-order functions (functions inside functions)

```python
def greetings():
    print('Hello World')
```

```python
greetings()
```

```python
def convert_to_uppercase(func):
    def inner():
        result = func() # Calls the original function
        return result.upper() #Modify the result
    return inner # Return the modified function
```

```python
@convert_to_uppercase

def greet():
    return 'hello world'

greet()
```
