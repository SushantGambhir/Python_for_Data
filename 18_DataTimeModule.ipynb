```python
import datetime
```

```python
current_date_time=datetime.datetime.now()
```

```python
print(current_date_time)
print(type(current_date_time))
```

```python
current_date_day = datetime.datetime.today()
print(current_date_day)
print(type(current_date_day))
```

```python
yr = current_date_day.year
print(yr)
print(type(yr))
```

```python
mn = current_date_day.month
print(mn)
print(type(mn))
```

```python
# Similarly we can get day, hour, minute, seconds, milliseconds etc
```

```python
d = datetime.datetime(2001,8,31)
print(d)
print(d.month)
```

```python
d1 = datetime.date(2022,6,21)
d2 = datetime.date(2022,3,15)
print(d1-d2)
print(type(d1-d2))
```

```python
d3 = d1 + datetime.timedelta(10) # Increase by 10 seconds
print(d3)
```

```python
d4 = d3 + datetime.timedelta(hours=8)
print(d4)
```

```python
print(d.timestamp())

# timestamp() returns number of seconds from 1st January 1970
```
